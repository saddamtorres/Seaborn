### 0.1 — La identidad visual del canal, tal como vivía en EP03.

```python
# Identidad visual del canal: los mismos colores de EP03
NAVY   = "#0A192F"   # fondo de la figura (Figure)
PANEL  = "#112B47"   # fondo de los ejes (Axes), distinto a propósito
CYAN   = "#00E5FF"   # valores medidos
ORANGE = "#FF9F1C"   # resultados / ajustes
GRID   = "#274764"   # rejilla y bordes
TEXTO  = "#E8EEF4"   # títulos, etiquetas y ticks
LINEA  = "#A9B8C9"   # bigotes, medianas y barras de error

# Un color fijo por elemento, el mismo en todo el canal
COLOR_EL = {"Au": "#FFC300", "Ag": "#C0C8D0", "Pb": "#8E9AAF",
            "Zn": "#00E5FF", "Cu": "#E07A5F", "Fe": "#D9622B"}

# Antes de tocar nada: matplotlib trae sus valores de fábrica
print("figure.facecolor ANTES :", plt.rcParams["figure.facecolor"])
print("axes.facecolor   ANTES :", plt.rcParams["axes.facecolor"])
```

### 0.2 — Lo mismo, pero delegado a seaborn.

```python
# Una sola llamada reemplaza el bloque de rcParams de EP03
sns.set_theme(
    style="darkgrid",        # estilo base de seaborn: fondo con rejilla
    context="notebook",      # escala de textos y líneas
    rc={                     # rc = el MISMO diccionario de matplotlib, con nuestros valores
        # --- colores: fondos, rejilla y texto
        "figure.facecolor": NAVY,
        "axes.facecolor":   PANEL,
        "savefig.facecolor": NAVY,     # para que el PNG exportado conserve el fondo
        "grid.color":       GRID,
        "text.color":       TEXTO,
        "axes.labelcolor":  TEXTO,
        "xtick.color":      TEXTO,
        "ytick.color":      TEXTO,
        "axes.edgecolor":   GRID,
        "figure.figsize":   (10, 6),  # tamaño por defecto de cada figura
        # tipografía: títulos 16, ejes 12, ticks y leyendas 10
        "axes.titlesize":   16,
        "figure.titlesize": 16,
        "axes.labelsize":   12,
        "xtick.labelsize":  10,
        "ytick.labelsize":  10,
        "legend.fontsize":  10,
        "legend.title_fontsize": 11,
        "font.size":        11,
    },
)

# La prueba: set_theme escribió en plt.rcParams, el mismo diccionario global de EP03
print("figure.facecolor DESPUÉS:", plt.rcParams["figure.facecolor"])
print("axes.facecolor   DESPUÉS:", plt.rcParams["axes.facecolor"])
```
